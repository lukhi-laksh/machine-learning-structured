```python
import numpy as np
```

```python
# Linear Regression
class LinearRegression:
    def __init__(self, learning_rate=0.01, n_itr=20000):
        self.bias = None
        self.weights = None
        self.lr = learning_rate
        self.n_itr = n_itr

    def fit(self, X, y): # X_train, y_train
        m, n = X.shape # (Number of sample, number of features)

        # Step 1 - Initalize parameter
        self.bias = 0
        self.weights = np.zeros(n) # Return array which have N size and all initilize at 0

        # Step 4
        # Gradient Descent
        for i in range(self.n_itr):
            # Step 2 - Calculate y_pred
            y_pred = self.bias + np.dot(X, self.weights)
    
            # Step 3 - Calculate Gradients
            db = (1/m) * np.sum(y_pred - y)
            dw = (1/m) * np.dot(X.T, (y_pred - y))

            # Step 4 - convergence therom - Parameter Upgrade
            self.bias -= self.lr * db
            self.weights -= self.lr * dw

    def predict(self, X):
        return self.bias + np.dot(X, self.weights)
```

```python
X = np.array([[1], [2], [3], [4], [5]])
y = np.array([2, 4, 6, 8, 10])

model = LinearRegression()
model.fit(X, y)

y_pred = model.predict(X)
print(y_pred)
```
