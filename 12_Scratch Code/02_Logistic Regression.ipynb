```python
import numpy as np
```

```python
# Logistic Regression

class LogisticRegression:
    def __init__(self, learning_rate = 0.01, n_itr = 10000):
        self.bios = None
        self.weight = None
        self.lr = learning_rate
        self.n_itr = n_itr
        
    def _sigmoid(self, z):
        return (1 / (1 + np.exp(-z)))


    def fit(self, X, y):
        m, n = X.shape
        self.bios = 0
        self.weight = np.zeros(n)

        for i in range(self.n_itr):
            z = self.bios = np.dot(X, self.weight)
            y_pred = self._sigmoid(z)

            db = (1/m) * (y_pred - y)
            dw = (1/m) * np.dot(X.T, (y_pred - y))

            self.bios -= self.lr * db
            self.weight -= self.lr * dw

    def get_probabilitys(self, X):
        z = self.bios = np.dot(X, self.weight)
        return self._sigmoid(z)

    def predict(self, X, threshold=0.5):
        probability = self.get_probabilitys(X)
        y_pred_bool = probability >= threshold
        return y_pred_bool.astype(int)
```

```python
X = np.array([[1, 2], [2, 3], [3, 4], [4, 5]])
y = np.array([0, 0, 1, 1])

model = LogisticRegression()
model.fit(X, y)

y_pred = model.predict(X)
y_pred
```
